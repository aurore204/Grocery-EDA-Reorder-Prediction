Définition des segments de clients

Objectif : distinguer les clients occasionnels, typiques et fidèles, pour orienter nos analyses et recommandations business.

Méthodologie : On va s'appuiyer sur la distribution réelle du nombre de commandes par utilisateur, mesurée avec .describe() :

```python
orders_per_user = orders.groupby("user_id")["order_number"].max()
orders_per_user.describe()
```

Observation : la moyenne (16,6) est nettement supérieure à la médiane (10), ce qui indique une minorité de très gros clients (jusqu'à 100 commandes) tire la moyenne vers le haut, alors que la majorité des utilisateurs commande beaucoup moins.

Segmentation retenue, basée sur les quartiles (Q1 et Q3) :

Occasionnels : ≤ 6 commandes (25% des utilisateurs les moins actifs, sous Q1)
Typiques : entre 7 et 20 commandes (50% central des utilisateurs, entre Q1 et Q3)
Fidèles : > 20 commandes (25% des utilisateurs les plus actifs, au-dessus de Q3)

Cette approche a l'avantage d'être directement ancrée dans les données observées, plutôt qu'un choix arbitraire, les seuils reflètent la structure réelle du comportement d'achat de la base client.
